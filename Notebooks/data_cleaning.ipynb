# Data Cleaning

## Data Loading

```python
# library imports
import pandas as pd
```

```python
# invoices
invoices = pd.read_csv('../Data/Original/invoices.csv')
invoices.head()
```

```python
# invoice charges
invoice_charges = pd.read_csv('../Data/Original/invoice_charges.csv')
invoice_charges.head()
```

```python
# payments
payments = pd.read_csv('../Data/Original/payments.csv')
payments.head()
```

```python
# projects
projects = pd.read_csv('../Data/Original/projects.csv')
projects.head()
```

```python
# vendors
vendors = pd.read_csv('../Data/Original/vendors.csv')
vendors.head()
```

## Data Cleaning

### Invoices

```python
# total invoices count
print('INVOICES:')
print('Total count:')
print(invoices.count(), '\n')

# check datatypes
print('Data type:')
print(invoices.dtypes, '\n')

# Check total missing values per column
print('Missing values:')
print(invoices.isna().sum(), '\n')

# Check for duplicates
print('Duplicate values:')
print(invoices.duplicated().sum(), '\n')

# missing values in approved_date are expected
```

```python
# convert dates to datetime
date_cols = ['invoice_date', 'received_date', 'due_date', 'approved_date']
invoices[date_cols] = invoices[date_cols].apply(pd.to_datetime, format='mixed' , errors='coerce')

# Apply appropriate datatypes
invoices = invoices.astype({
    'invoice_id': 'string',
    'vendor_id': 'string',
    'invoice_number': 'string',
    'approval_status': 'category'
})

invoices.dtypes
```

### Invoice Charges

```python
# invoice_charges count
print('INVOICE_CHARGES:')
print('Total count:')
print(invoice_charges.count(), '\n')

# check datatypes
print('Data type:')
print(invoice_charges.dtypes, '\n')

# Check total missing values per column
print('Missing values:')
print(invoice_charges.isna().sum(), '\n')

# Check for duplicates
print('Duplicate values:')
print(invoice_charges.duplicated().sum(), '\n')
```

```python
# Apply appropriate datatypes
string_cols = ['charge_id', 'invoice_id', 'project_id']
invoice_charges[string_cols] = invoice_charges[string_cols].astype('string')

invoice_charges['charge_type'] = invoice_charges['charge_type'].astype('category')

invoice_charges.dtypes
```

### Payments

```python
# payments count
print('PAYMENTS:')
print('Total count:')
print(payments.count(), '\n')

# check datatypes
print('Data type:')
print(payments.dtypes, '\n')

# Check total missing values per column
print('Missing values:')
print(payments.isna().sum(), '\n')

# Check for duplicates
print('Duplicate values:')
print(payments.duplicated().sum(), '\n')
```

```python
# Apply appropriate datatypes
payments['payment_date'] = payments['payment_date'].apply(pd.to_datetime, format='mixed' , errors='coerce')

payments = payments.astype({
    'payment_id': 'string',
    'invoice_id': 'string',
    'payment_method': 'category',
    'payment_status': 'category'
})

payments.dtypes
```

### Projects

```python
# projects count
print('PROJECTS:')
print('Total count:')
print(projects.count(), '\n')

# check datatypes
print('Data type:')
print(projects.dtypes, '\n')

# Check total missing values per column
print('Missing values:')
print(projects.isna().sum(), '\n')

# Check for duplicates
print('Duplicate values:')
print(projects.duplicated().sum(), '\n')

# missing end date values are expected
```

```python
# Apply appropriate datatypes
projects['start_date'] = projects['start_date'].apply(pd.to_datetime, format='mixed' , errors='coerce')
projects['end_date'] = projects['end_date'].apply(pd.to_datetime, format='mixed' , errors='coerce')

projects = projects.astype({
    'project_id': 'string',
    'project_name': 'string',
    'client_name': 'string',
    'project_type': 'category',
    'project_status': 'category',
    'project_manager': 'string',
    'budget_amount': 'float64'
    
})

projects.dtypes
```

### Vendors

```python
# vendors count
print('VENDORS:')
print('Total count:')
print(vendors.count(), '\n')

# check datatypes
print('Data type:')
print(vendors.dtypes, '\n')

# Check total missing values per column
print('Missing values:')
print(vendors.isna().sum(), '\n')

# Check for duplicates
print('Duplicate values:')
print(vendors.duplicated().sum(), '\n')
```

```python
# Apply appropriate datatypes
vendors = vendors.astype({
    'vendor_id': 'string',
    'vendor_name': 'string',
    'vendor_type': 'category',
    'payment_terms': 'category',
    'active_flag': 'category'
    
})

vendors.dtypes
```

## Data Exports

```python
# Cleaned datasets were saved as pickle files locally to preserve pandas data types between notebooks. 
# CSV versions are included in the repository for transparency.

# export cleaned data to pickle to preserve data types
invoices.to_pickle('../Data/Cleaned/invoices_cleaned.pkl')
invoice_charges.to_pickle('../Data/Cleaned/invoice_charges_cleaned.pkl')
payments.to_pickle('../Data/Cleaned/payments_cleaned.pkl')
projects.to_pickle('../Data/Cleaned/projects_cleaned.pkl')
vendors.to_pickle('../Data/Cleaned/vendors_cleaned.pkl')

# export cleaned data to csv
invoices.to_csv('../Data/Cleaned/invoices_cleaned.csv')
invoice_charges.to_csv('../Data/Cleaned/invoice_charges_cleaned.csv')
payments.to_csv('../Data/Cleaned/payments_cleaned.csv')
projects.to_csv('../Data/Cleaned/projects_cleaned.csv')
vendors.to_csv('../Data/Cleaned/vendors_cleaned.csv')
```
